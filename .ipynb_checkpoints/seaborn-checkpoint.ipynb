# Seaborn Tutorial

### Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
%reload_ext autoreload
```

## Import data

```python
df = pd.read_csv('car_crashes.csv')
tips_df = pd.read_csv('tips.csv')
```

## Distribution plots

### Distribution Plot

```python
# sns.displot(df['not_distracted'], kde=False, bins = 25 )
# plt.show()
df
```

### Joint plot

```python
sns.jointplot(x='speeding', y='alcohol', data = df, kind='hex')
plt.show()
```

### Kde Plot

```python
sns.kdeplot(df['alcohol'])
plt.show()
```

### Pairplot

```python
sns.pairplot(df)
plt.show()
```

### Rugplot

```python
sns.rugplot(df['alcohol'])
plt.show()
```

### Styling

```python
sns.set_style('ticks')

# plt.figure(figsize=(8,4))

sns.set_context('paper', font_scale=2.0)

sns.jointplot(x='speeding', y='alcohol', data = df, kind='reg')

# sns.despine(left=True, bottom=True)
plt.show()
```

## Categorical plots

### Bar plots

```python
tips_df
```

### Count plot

```python
sns.countplot(x='sex', data=tips_df)
plt.show()
```

### Box plot

```python
sns.boxplot(x='day', y='total_bill', data=tips_df, hue='sex' )
plt.show()
```

## Violin Plot

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex', split=True )
plt.show()
```

### Strip plot

```python
plt.figure(figsize=(8,5))

sns.stripplot(x='day', y='total_bill', data=tips_df, hue='sex', dodge=True)
plt.show()
```

### Swarm plot

```python
sns.violinplot(x='day', y='total_bill', data=tips_df, hue='sex', dodge=True)
sns.swarmplot(x='day', y='total_bill', data=tips_df, hue='sex', dodge=True, palette='dark:white')
plt.show()
```

## Palettes

```python
plt.figure(figsize=(8,5))
sns.set_style('dark')
sns.set_context('talk')
sns.stripplot(x='day', y='total_bill', data=tips_df, hue='sex', palette='magma')
plt.legend(loc=[1,.1])
plt.show()
```
